```python
from pathlib import Path

from d3rlpy.dataset import InfiniteBuffer, ReplayBuffer


EXPERT_DATASET_PATH = Path("../outputs/mdp_dataset_trajectories_expert_5k.h5")
RANDOM_DATASET_PATH = Path("../outputs/mdp_dataset_trajectories_random_5k.h5")
TWO_POLICIES_DATASET_PATH = Path("../outputs/mdp_dataset_trajectories_two_policies_5k.h5")

mdp_dataset_expert = ReplayBuffer.load(EXPERT_DATASET_PATH.open("rb"), InfiniteBuffer())
mdp_dataset_random = ReplayBuffer.load(RANDOM_DATASET_PATH.open("rb"), InfiniteBuffer())
for episode in mdp_dataset_random.episodes:
       mdp_dataset_expert.append_episode(episode)
       print(f"Merged episodes: {mdp_dataset_expert.size()}")

    # Do not overwrite either source dataset.
with TWO_POLICIES_DATASET_PATH.open("w+b") as f:
    mdp_dataset_expert.dump(f)
```
